```python
import numpy as np # linear algebra
import pandas as pd # data processing, CSV file reading and writing
from sklearn import preprocessing
from sklearn.model_selection import train_test_split
```

```python
datasets = "./datasets/datasets - crop irrigation.csv"
raw_data = pd.read_csv(datasets)
raw_data.head()
```

```python
raw_data.info()
```

```python
X = raw_data.drop('Irrigation', axis="columns")
Y = raw_data['Irrigation']
X["CropType"].unique()
```

```python
label_encoder = preprocessing.LabelEncoder()
X['CropType'] = label_encoder.fit_transform(X['CropType']) # encode the crop type label
X['CropType'].unique()
```

```python
label_encoder.fit_transform(["Paddy", "Coffee", "Groundnuts", "Maize", "Wheat", "Garden Flowers", "Potato", "Sugarcane", "Pulse"])
```

```python
Xtrain, Xtest, Ytrain, Ytest = train_test_split(X,Y,train_size = 0.8)
```

```python
Xtrain.shape
```

```python
np.savez('./processed_data/crop_data.npz', Xtrain=Xtrain, Ytrain=Ytrain, Xtest=Xtest, Ytest=Ytest)
```
