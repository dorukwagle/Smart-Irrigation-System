```python
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import accuracy_score, classification_report
import numpy as np
```

```python
data = np.load("./processed_data/crop_data.npz")
Xtrain = data["Xtrain"]
Ytrain = data["Ytrain"]
Xtest = data["Xtest"]
Ytest = data["Ytest"]
Xtrain.shape
```

```python
# initialize the model
mlp = MLPClassifier(hidden_layer_sizes=(10,), activation='relu', solver='adam', max_iter=100, learning_rate="adaptive", random_state=42, n_iter_no_change=10, verbose=True)

# Train the model
mlp.fit(Xtrain, Ytrain)
```

```python
y_pred = mlp.predict(Xtest)
```

```python
print("Accuracy:", accuracy_score(Ytest, y_pred))
```
